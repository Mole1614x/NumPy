# Basic

```python
import numpy as np
```

```python
# Q1. Create an array from 1 to 10

arr = np.arange(1,11)
print(arr)
```

```python
# Q2. Create an 3X3 matrix of zeroes

arr = np.zeros((3,3))
print(arr)
```

```python
# Q3. Create an array of even numbers from 2 to 20.

arr = np.arange(2,21,2)
print(arr)
```

```python
# Q4. Find shape,size and dimensions of a 2D array.

arr_2d = np.array([
    [1,2],
    [4,5]
])


print("Shape:", arr_2d.shape)
print("Size:", arr_2d.size)
print("Dimension:", arr_2d.ndim)
```

# Intermediate

```python
# Q6. Create a 3 X 4 array and reshape it into 4 X 3.

arr = np.arange(1,13).reshape(3,4)
print(f"Original 3 X 4 array:")
print(arr)

reshaped_array = arr.reshape(4,3)
print(reshaped_array)
```

```python
# Q7. Find row-wise and column-wise sum.

arr = np.array([
    [1,2],
    [5,8]
])
print(np.sum(arr))
print(f"The row-wise sum is: {np.sum(arr, axis=1)}")
print(f"The Column-wise sum is: {np.sum(arr,axis=0)}")
```

```python
# Q8. Filter. all values greater than 50.

arr = np.array([45,67,89,32,76,90,55])

print(arr[arr > 50])
```

```python
# Q9. Replace values less than 40 with 0

arr = np.array([10, 25, 40, 55, 30, 70])

arr[arr < 40] = 0
print(arr)
```

```python
# Q10. FIND UNIQUE VALUES AND THEIR COUNTS

arr = np.array([1,2,3,2,4,4,2,3,5])
print(np.unique(arr))

value,counts = np.unique(arr,return_counts=True)
print(f"The Value is: {value}, The Count is: {counts}")
```

# Advanced

```python
# Q11. Perform matrix multiplication of two 2 X 2 matrices.

arr1 = np.arange(1,5).reshape(2,2) 
arr2 = np.arange(9,13).reshape(2,2)
print(arr1)
print(arr2)

result = arr1 @ arr2
multiply = np.matmul(arr1,arr2)
mat_mul = np.dot(arr1,arr2)
print(result)
print(multiply)
print(mat_mul)
```

```python
# Q12. Find Determinant and inverse of a matrix

arr = np.array([
    [1,2],
    [3,4]
])

# Determinant

print(f"Determinant: {np.linalg.det(arr)}")

# Inverse

print(f"Inverse: {np.linalg.inv(arr)}")
```

```python
# Q13. Generate random marks of 10 students and classify Pass/Fail

marks = np.random.randint(0,101,10)

result = np.where(marks >= 40 ,"Pass","Fail")
print("Marks:", marks)
print("Result:", result)
```

```python
# Q14. Find Q1,Q3 and IQR of an array

arr = np.array([10,20,30,40,50,60,70,80])
q1 = np.percentile(arr,25)
q3 = np.percentile(arr,75)
iqr = q3 - q1

print("Q1:",q1)
print("Q3:",q3)
print("IQR:",iqr)
```

```python
# Q15. Use broadcasting to add a 1D array to every row of a 2D array

arr = np.arange(1,13).reshape(3,4)
print(arr)
arr1 = np.arange(1,5)
print(arr1)
result = arr + arr1
print(result)
```
